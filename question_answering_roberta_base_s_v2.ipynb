```python
import requests
from bs4 import BeautifulSoup
from transformers import pipeline

def scrape_wikipedia_page(url):
  """Scrapes the content of a Wikipedia page.

  Args:
    url: The URL of the Wikipedia page.

  Returns:
    A string containing the text content of the page, or None if an error occurs.
  """
  try:
    response = requests.get(url)
    response.raise_for_status()  # Raise an exception for bad status codes
    soup = BeautifulSoup(response.content, 'html.parser')

    # Extract the main content from the page (excluding infoboxes, sidebars, etc.)
    content = ""
    for paragraph in soup.select('div.mw-parser-output p'):
        content += paragraph.get_text() + "\n"

    return content

  except requests.exceptions.RequestException as e:
    print(f"Error fetching the page: {e}")
    return None

# Example usage:
wikipedia_url = "https://en.wikipedia.org/wiki/India"
page_content = scrape_wikipedia_page(wikipedia_url)

question = "Which is the official name of india?"

# Reuse the pipeline from the previous cell
question_answerer = pipeline("question-answering", model="consciousAI/question-answering-roberta-base-s-v2")
answer = question_answerer(question=question, context=page_content)
print(answer)
```
